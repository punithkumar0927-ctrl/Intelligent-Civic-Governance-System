# NEXUS AI - 02 Image PreprocessingCLAHE, denoising, sharpening, and augmentation pipeline.

```python
import cv2, numpy as np, matplotlib.pyplot as pltfrom pathlib import Pathimport sys; sys.path.insert(0, str(Path.cwd().parent))from ai_engine.detection.preprocessing import preprocess_image, PreprocessConfigcfg = PreprocessConfig()samples = list(Path('../datasets/road_damage/images/train').glob('*.jpg'))print('Config:', cfg); print('Samples found:', len(samples))
```

```python
if samples:    img = cv2.imread(str(samples[0]))    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)    processed = preprocess_image(str(samples[0]), cfg)    processed_rgb = cv2.cvtColor(processed, cv2.COLOR_BGR2RGB)    fig, axes = plt.subplots(1, 2, figsize=(12,5))    axes[0].imshow(img_rgb); axes[0].set_title('Original')    axes[1].imshow(processed_rgb); axes[1].set_title('Preprocessed (CLAHE+Sharpen)')    for ax in axes: ax.axis('off')    plt.tight_layout(); plt.show()
```

```python
if samples:    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)    edges_raw = cv2.Canny(gray, 50, 150)    gray_p = cv2.cvtColor(processed, cv2.COLOR_BGR2GRAY)    edges_clahe = cv2.Canny(gray_p, 50, 150)    fig, axes = plt.subplots(2, 2, figsize=(12,8))    axes[0,0].imshow(img_rgb); axes[0,0].set_title('Original')    axes[0,1].imshow(processed_rgb); axes[0,1].set_title('CLAHE Enhanced')    axes[1,0].imshow(edges_raw, cmap='gray'); axes[1,0].set_title('Edges (raw)')    axes[1,1].imshow(edges_clahe, cmap='gray'); axes[1,1].set_title('Edges (CLAHE)')    for ax in axes.flat: ax.axis('off')    plt.tight_layout(); plt.show()
```
