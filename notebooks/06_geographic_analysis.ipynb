```python
import sysfrom pathlib import Pathsys.path.insert(0, str(Path.cwd().parent))import pandas as pdimport matplotlib.pyplot as pltfrom datetime import datetime, timezonefrom ai_engine.geographic_intelligence.hotspot_analysis import find_hotspots, IssuePoint
```

```python
df = pd.read_csv('../datasets/geographic_intelligence/issue_coordinates.csv')print(len(df), 'issues loaded'); df.head()
```

```python
points = []for _, row in df.iterrows():    points.append(IssuePoint(        issue_id=int(row['issue_id']),        lat=float(row['lat']), lon=float(row['lon']),        severity=str(row.get('severity', 'medium')),        category=str(row.get('category', 'other')),        status='open',        created_at=datetime.now(timezone.utc),    ))print(len(points), 'points ready')
```

```python
hotspots = find_hotspots(points, eps_km=0.3, min_pts=2)for h in hotspots:    print(h.hotspot_id, h.issue_count, 'issues', 'risk=', round(h.risk_score, 3), h.top_category, 'trend=', h.trend)
```

```python
try:    import folium    m = folium.Map(location=[12.9716, 77.5946], zoom_start=11, tiles='CartoDB positron')    sev_c = {'LOW':'green','MEDIUM':'orange','HIGH':'red','CRITICAL':'purple'}    for p in points:        folium.CircleMarker([p.lat, p.lon], radius=8,            color=sev_c.get(p.severity.upper(), 'gray'), fill=True,            popup=p.category + ' | ' + p.severity).add_to(m)    for h in hotspots:        folium.Circle([h.centroid_lat, h.centroid_lon],            radius=int(h.radius_km * 1000), color='crimson',            fill=True, fill_opacity=0.15,            popup='Hotspot: ' + str(h.issue_count) + ' issues').add_to(m)    mexcept ImportError:    print('Install folium: pip install folium')
```
