```python
import cv2, numpy as np, matplotlib.pyplot as pltfrom pathlib import Pathimport sys; sys.path.insert(0, str(Path.cwd().parent))from ai_engine.resolution_verification.damage_comparison import (    ssim_single, edge_density, region_comparison)
```

```python
# Simulate before/after pairnp.random.seed(42)before = np.random.randint(80, 200, (300, 400, 3), dtype=np.uint8)after = before.copy()mask = np.zeros((300, 400), dtype=bool)mask[100:200, 150:250] = Trueblurred = cv2.GaussianBlur(after, (15, 15), 0)after[mask] = blurred[mask]fig, axes = plt.subplots(1, 2, figsize=(12, 5))axes[0].imshow(cv2.cvtColor(before, cv2.COLOR_BGR2RGB)); axes[0].set_title('Before')axes[1].imshow(cv2.cvtColor(after, cv2.COLOR_BGR2RGB)); axes[1].set_title('After')for ax in axes: ax.axis('off')plt.tight_layout(); plt.show()
```

```python
score = ssim_single(before, after)eb = edge_density(before); ea = edge_density(after)reg = region_comparison(before, after)print('SSIM:', round(score, 4))print('Edge before:', round(eb, 4), 'Edge after:', round(ea, 4))print('Edge reduction:', reg['edge_reduction_pct'], '%')
```
