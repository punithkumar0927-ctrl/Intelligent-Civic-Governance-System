# NEXUS AI - 01 Dataset ExplorationExplore RDD2022 road damage dataset structure, class distribution, and image quality.

```python
import os, sysfrom pathlib import Pathimport matplotlib.pyplot as pltfrom PIL import Imageimport numpy as npsys.path.insert(0, str(Path.cwd().parent))DATASET = Path('../datasets/road_damage')print('Dataset exists:', DATASET.exists())print('Images in train:', len(list((DATASET/'images/train').glob('*.jpg'))))print('Labels in train:', len(list((DATASET/'labels/train').glob('*.txt'))))
```

```python
from collections import Counterclass_counts = Counter()for lbl in (DATASET/'labels/train').glob('*.txt'):    for line in lbl.read_text().splitlines():        parts = line.strip().split()        if parts: class_counts[int(float(parts[0]))] += 1CLASSES = ['medium_damaged','severely_damaged','unpaved','longitudinal_transverse_crack','alligator_crack']names = [CLASSES[i] for i in sorted(class_counts)]plt.bar(names, class_counts.values(), color='steelblue')plt.title('RDD2022 Class Distribution'); plt.xticks(rotation=20); plt.tight_layout(); plt.show()
```

```python
imgs = list((DATASET/'images/train').glob('*.jpg'))[:4]fig, axes = plt.subplots(1, 4, figsize=(16,4))for ax, img_path in zip(axes, imgs):    img = Image.open(img_path); ax.imshow(img)    ax.set_title(str(img.size), fontsize=9); ax.axis('off')plt.suptitle('Sample Images'); plt.tight_layout(); plt.show()
```
